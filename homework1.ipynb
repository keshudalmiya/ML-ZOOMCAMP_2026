```python
X=df_asia.to_numpy()
```

```python

# Calculate the transpose of X and XTX
XT = X.T
XTX = np.dot(XT, X)
XTX
```

```python
# Invert XTX
XTX_inverse = np.linalg.inv(XTX)
XTX_inverse
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = np.dot(np.dot(XTX_inverse, XT), y)
w
```

```python
sum_w = np.sum(w)
sum_w
```
